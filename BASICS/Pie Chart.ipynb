<p style="text-align:center; font-size:30px;"><b>PIE CHART</b></p>

<span style="font-size:20px;font-family:Arial;">Pie chart is a circular chart divided into slices to show percentages of the total.
good for vistualization distribution among categories</span>

```python
import matplotlib.pyplot as plt
import numpy as np
```

<span style="font-size:20px;">**Creation of Pie chart**</span>

```python
categories=["Freshman","Sophomores","Juniors","Seniors"]
values=np.array([300,275,225,100])
plt.pie(values)
plt.show()
```

<p style=" font-size:20px;"><b>Adding labels to the chart</b></p>

```python
categories=["Freshman","Sophomores","Juniors","Seniors"]
values=np.array([300,275,225,100])
plt.pie(values,labels=categories)      #Assigning categories to the labels
plt.show()
```

<p style=" font-size:20px;"><b>Adding Custom colors to the chart</b></p>

```python
categories=["Freshman","Sophomores","Juniors","Seniors"]
values=np.array([300,275,225,100])
colors=["yellow","blue","green","red"]   #Custom colors
plt.pie(values,labels=categories,
                      colors=colors)
plt.show()
```

<p style=" font-size:20px;"><b>Adding Auto percentage to the chart</b></p>

```python
categories=["Freshman","Sophomores","Juniors","Seniors"]
values=np.array([300,275,225,100])
colors=["yellow","blue","green","red"]
plt.pie(values,labels=categories,
                      colors=colors,
                      autopct="%1.1f%%")       #autopct is used to automatically display the percentage value on each slice of the pie.
plt.show()
"""
Example For autopct:----

Changing decimal places:
No decimal:
    autopct="%1.0f%%"
    autopct="%1.0f%"
Output:
    50%,30%,20%
    5%,3%,2%

Two decimals:
    autopct="%1.2f%%"
    autopct="%1.2f%"
Output:
    50.00%,60.66%,45.98%
    5.00%,6.06%,4.59%

"""
```

<p style=" font-size:20px;"><b>Other parameters to the pie chart</b></p>

```python
categories=["Freshman","Sophomores","Juniors","Seniors"]
values=np.array([300,275,225,100])
colors=["yellow","blue","green","red"]
plt.pie(values,labels=categories,
                      colors=colors,
                      autopct="%1.1f%%",
                      explode=[0,0,0,0.2], #separate a slice (larger the value,further the distance that pi slice will explode
                      shadow=True,         #add shadow effect
                      startangle=90)       #rotate pie chart
plt.title("Pie Chart")
plt.show()
```
