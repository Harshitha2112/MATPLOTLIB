<h1 style="text-align:center;"><b>Box Plot</b></h1>

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
"""
A box plot summarizes data using:

    Minimum value
    Q1 (25%)
    Median (50%)
    Q3 (75%)
    Maximum value
    Outliers

Used for:
    Finding outliers
    Understanding spread
"""

data=[10,20,30,40,50,100]

plt.boxplot(data)
plt.show()
```

<h1 style="text-align:center;"><b>Styling And Fill Between</b></h1>

```python
x=np.arange(10)
y1=x
y2=x+5
plt.style.use("ggplot")  # changes the plot appearence
#plt.style.use("dark_background")
plt.fill_between( x, y1, y2)
plt.show()


"""
The three parameters are:
x → X-axis values
y1 → Lower boundary
y2 → Upper boundary
It fills the area between y1 and y2.
ex:
    Suppose a machine learning model predicts accuracy:
    epochs = [1,2,3,4,5]
    accuracy = [70,75,80,85,90]
    But accuracy has uncertainty:
        Accuracy ± 5%
So:
Lower limit:
lower = [65,70,75,80,85]

Upper limit:
upper = [75,80,85,90,95]
The shaded region shows the possible range.
"""
```

<h1 style="text-align:center;"><b>Histogram</b></h1>

<h1 style="font-family:Arial;">A histogram shows the distribution of numerical data.</h1>


```python
marks = np.random.normal(70,10,1000)

plt.hist(
    marks,
    bins=20,
    density=True
)

plt.xticks(rotation=45)
plt.show()

"""
Normally histogram shows:
Number of observations in each bin"

Marks:-
50-60  █████  5 students
60-70  █████████ 9 students
70-80  ████████████ 12 students
it converts the y-axis into probability density.
The total area of all bars becomes 1.

ex:
    Imagine comparing:
    Class A:
    100 students
    Class B:
    1000 students
    
    Normal histogram is unfair because Class B has more students.
    Density converts both into a common scale.

Rotating X labels
When labels overlap:
"""
```

<h1 style="text-align:center;"><b>Errorbar Plot</b></h1>

<h1 style="font-size:20px;font-family:Arial;">Shows uncertainty or error.</h1>

```python
x=[1,2,3]
y=[80,85,90]
error=[5,3,4]

plt.errorbar(
    x,
    y,
    yerr=error,
    fmt='o'
)
plt.show()
"""
Suppose 5 students have average marks:  Average = 80
But actual marks vary:  78, 82, 79, 85, 76
So saying only: 80 is incomplete.

We need uncertainty: 80 ± 5
This is an error bar.
The vertical line shows possible variation.
yerr is the variation of the particular data points such as 80 varies by + or - 5 
"""
```

<h1 style="text-align:center;"><b>Axis Limits</b></h1>

<h1 style="font-size:20px;font-family:Arial;">it controls the visible range</h1>

```python
plt.plot([1,2,3,4],[10,20,30,40])
plt.show()

plt.plot([1,2,3,4],[10,20,30,40])
plt.xlim(1,3)
plt.ylim(10,30)

plt.show()


"""
Example:
Data:
    x=[1,2,3,4]
    y=[10,20,30,40]

Normally:
    x-axis:
    0 to 4
    y-axis:
    0 to 40

You can change:
    plt.xlim(1,3)
    plt.ylim(10,30)

Now only:
    x = 1 to 3
    y = 10 to 30 is visible.
"""
```

<h1 style="text-align:center;"><b>Annotate</b></h1>

<h1 style="font-size:20px;font-family:Arial;">Add a note pointing to a specific point.</h1>

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 25, 18, 35, 30]

plt.plot(x, y, marker='o')

plt.annotate(
    "Highest Point",                    # Text to display
    xy=(4, 35),                         # Point being highlighted
    xytext=(3, 40),                     # Where the text appears
    arrowprops=dict(arrowstyle="->",    ## Arrow style
                   color="blue",
                   linewidth=2)
)

plt.ylim(0, 45)      # Increase y-axis limit beacause text appearing outside the graph cause y values has till 30 only

plt.title("Annotate Example")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

<h1 style="text-align:center;"><b>Text</b></h1>

<h1 style="font-size:20px;font-family:Arial;">Adds text anywhere.</h1>

```python
#label single point

x = [1, 2, 3, 4]
y = [10, 20, 30, 25]

plt.plot(x, y, marker='o')
plt.text(2, 22, "Hello")
plt.show()

#Label Every Point

x = [1, 2, 3, 4]
y = [10, 20, 30, 25]

plt.plot(x, y, marker='o')
for i in range(len(x)):
    plt.text(x[i], y[i], f"({x[i]},{y[i]})")

plt.show()
```

<h1 style="text-align:center;"><b>Twin Axis</b></h1>

<h1 style="font-size:20px;font-family:Arial;">Creates two Y axes.<br>
Used when values have different scales.</h1>

```python

# Data
months = ["Jan", "Feb", "Mar", "Apr", "May"]

sales = [2000, 3500, 3000, 4500, 5000]
temperature = [22, 24, 28, 30, 32]

# Create first axis
fig, ax1 = plt.subplots()

# Left Y-axis
ax1.plot(months, sales, color="blue", marker="o", label="Sales")
ax1.set_xlabel("Months")
ax1.set_ylabel("Sales (₹)", color="blue")

# Create second Y-axis
ax2 = ax1.twinx()

# Right Y-axis
ax2.plot(months, temperature, color="red", marker="s", label="Temperature")
ax2.set_ylabel("Temperature (°C)", color="red")

plt.title("Sales vs Temperature")
plt.show()

"""
Suppose you want to compare:
Sales (₹1000–₹5000)
Temperature (20–35°C)

If you plot both on the same Y-axis, the temperature line will look almost flat because sales values are much larger.
A twin axis gives each dataset its own Y-axis.
Both use the same X-axis (Months) but different Y-axes.
"""
```

<h1 style="text-align:center;"><b>3D plotting</b></h1>

<h1 style="font-size:20px;font-family:Arial;">It is used when your data has three variables.</h1>

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Data
x = [1,2,3,4,5]
y = [5,6,2,8,7]
z = [10,15,12,20,18]

# Create figure
fig = plt.figure()

# Create 3D axis
ax = fig.add_subplot(projection='3d')

# Plot points
ax.scatter(x, y, z, color="red", s=100)

# Labels
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.title("3D Scatter Plot")

plt.show()
```
