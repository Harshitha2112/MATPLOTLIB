<h1 style="font-size:30px;text-align:center;"><b>Histogram</b></h1>

<h1 style="font-size:20px;font-family:Arial;">A visual representation of the distribution of quantitative data.
they group values into bins(intervals) and counts how many fall in each range.</h1>

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#np.random.normal  :--- This function generates random numbers from a normal distribution (also called a Gaussian distribution).
#this function takes 3 parameters i.e, loc,scale,size

scores = np.random.normal(loc=80,scale=10,size=100)


"""
loc :-
    loc is the mean (average) of the data.
    ex: loc = 80 ,This means the generated values will be centered around 80.

scale :-
    scale is the standard deviation.
    It tells you how spread out the numbers are.
    ex : scale = 10
        Small standard deviation → values stay close to 80.
        Large standard deviation → values spread farther from 80.

size :- 
    This tells NumPy how many random values to generate.
    ex: size = 100
        creates an array containing 100 random scores.

code creates 100 random student scores where the average score is about 80, and the scores are spread around that average by about 10 marks.
"""
```

```python
plt.hist(scores)
plt.show()
```

```python
scores = np.random.normal(loc=80,scale=100,size=100)
plt.hist(scores)
plt.show()
# if i set scale to big number, the values will go below zero or beyond the max value
#values will not be around mean value
```

<h1 style="font-size:30px;text-align:center;"><b>Clip</b></h1>

```python
#np.clip() limits the values to a specified range.
"""
Any value below 0 becomes 0.
Any value above 100 becomes 100.
Values between 0 and 100 stay unchanged.
"""
scores = np.random.normal(loc=80,scale=100,size=100)
scores=np.clip(scores,0,100)
plt.hist(scores,bins=10,            #bins is a number of bars in a chart
                color="lightgreen",
                edgecolor="Black")
plt.title("Exam Scores")
plt.xlabel("Score")
plt.ylabel("no of students")
plt.show()
```
