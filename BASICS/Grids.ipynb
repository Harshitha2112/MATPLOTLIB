<span style="font-size:20px;">**Grid helps make plots easier to read by addind refrence lines**</span> 

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x=np.array([1,2,3,4,5])
y=np.array([5,10,15,20,25])
plt.grid()
plt.plot(x,y)
plt.show()
```

```python
x=np.array([1,2,3,4,5])
y=np.array([5,10,15,20,25])
plt.grid(axis="x")        #only vertical lines are drawn
plt.plot(x,y)
plt.show()
```

```python
x=np.array([1,2,3,4,5])
y=np.array([5,10,15,20,25])
plt.grid(axis="y")         #only horizontal lines are drawn
plt.plot(x,y)
plt.show()
```

```python
x=np.array([1,2,3,4,5])
y=np.array([5,10,15,20,25])
plt.grid(color="lightgray",
         linewidth=2,
         linestyle="dashed")
plt.plot(x,y)
plt.show()
```
