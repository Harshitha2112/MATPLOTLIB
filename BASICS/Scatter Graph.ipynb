<h1 style="text-align:center;"><b>Scatter graph</b></h1>

<span style="font-size:20px;">**It shows relationship between two variables helps to identify a correlation(+,-,none)**</span>

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x1=[0,1,2,3,4,5,6,7,8]
y1=[50,55,60,65,70,80,90,95,99]

x2=[0,2,4,6,8,10,12,14,16]
y2=[56,55,60,63,70,89,90,45,99]


plt.scatter(x1,y1,color="skyblue",
                alpha=0.5,              #it controls the intensity of the color
                s=200,                  #it is used for size of the data points
                label="class A")        #label is to give name for particular class
plt.scatter(x2,y2,color="red",
                alpha=0.5,
                s=200,
                label="class B")

plt.legend()
plt.title("test scores")
plt.xlabel("hours studied")
plt.ylabel("grade")

plt.show()
```
