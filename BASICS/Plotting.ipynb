```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#it also works well with list but creating data as array is a best practice
x1=np.array([1,2,3,4,5])
y1=np.array([50,55,60,65,70])
plt.plot(x1,y1,marker='*',                # marker is to mark data points in the graph markers can be *,.,etc
             markersize=10,             # markersize is the size of the data points
             markerfacecolor="purple",  # markerfacecolor also written as mfc and is used to color the data points
             markeredgecolor="red",     # marheredgecolor also written as mec and is used to color the edges of the data points
             linewidth=4,               # linewidth is the width of the line
             linestyle="dashed",        # linestyle is the type of lines i.e, dashed,dotted,dashdot,solid,none
             color="green")             # color is the color of the line
plt.show()
```

```python
# now if we want same keyword arguments ,we could just copy all the keyword arguments from the previous graph and paste it in another graph
#but instead of it we can use dictionary to use same keyword arguments
x2=np.array([3,6,9,12,15])
y2=np.array([55,60,75,85,90])
#we create dictionary and pass all the keyword arguments
line_style=dict(marker='*',                
             markersize=10,             
             markerfacecolor="purple",  
             markeredgecolor="red",     
             linewidth=4,               
             linestyle="dashed",        
             color="green")
plt.plot(x2,y2,**line_style)    # we unpack the line_style dictionay by adding asterisks in front of it 
plt.show()
```

```python
#plotting both graphs in one graph
plt.plot(x1,y1,**line_style)
plt.plot(x1,y2,**line_style)

label=dict(fontsize=20,
           family="Arial",
           fontweight="bold",
          )

#adding title for the graph
plt.title("Class size", **label , color="blue")    # we customized the color of the labels and title

#customizing ticks
plt.tick_params(colors="brown",
               axis="both")

plt.xlabel("Year",**label, color="purple")    
plt.ylabel("Students",**label, color="purple")
plt.show()
```
