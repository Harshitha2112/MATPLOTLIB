<h1 style="text-align:center;"><b>Savefig()</b></h1>

<span style="font-size:20px;font-family:Arial;">it is a function for saving vistualization</span>

```python
import matplotlib.pyplot as plt
```

```python
#savefig('filename.extention,dpi=value,bbox_inches='tight')
"""
we can save visualization in different formats.
DPI = dots per inch
It controls image resolution.
For Data Science reports, 300 DPI(high quality) is commonly used.

bbox_inches="tight"
This removes unnecessary white space around the image.
"""
x=[1,2,3,4]
y=[10,20,30,25]

plt.plot(x,y,color='blue',marker='o')
plt.title("Simple line plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.savefig("line_plot.png",dpi=300,bbox_inches="tight")
plt.show()
```
