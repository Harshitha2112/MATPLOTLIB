<span style="font-size:20px;">**Bar chart compares categories of data by representing each category with a bar**</span>

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
categories=["protein","dairy","cereals","sweets","grains","fruits","vegetables"]
rating=np.array([10,7,6,1,4,8,6])
plt.bar(categories,rating)
plt.title("Daily consumption")
plt.xlabel("foods")
plt.ylabel("quantity")
plt.show()
```

<span style="font-size:20px;">**To plot graph in horizontal**<span>

```python
categories=["protein","dairy","cereals","sweets","grains","fruits","vegetables"]
rating=np.array([10,7,6,1,4,8,6])
plt.barh(categories,rating,color="pink")   #adding colors to the bars
plt.title("Daily consumption")
plt.xlabel("foods")
plt.ylabel("quantity")
plt.show()
```
