<h1 style="text-align:center;"><b>Subplots</b></h1>

<span style="font-size:20px;font-family:Arial;">figure is a entire canvas where we plot graphs. <br>Ax is a known as single subplot in a canvas.<br>A figure can contain multiple axes</span>

```python
#the subplot function will return a tuple. the tuple has 2 entities i.e, a figure which is our canvas and a 2-dimensional array of axes objects
#we are going to assign what's returned to us and unpack it.

import matplotlib.pyplot as plt
import numpy as np
print(plt.subplots(2,2))
```

```python
#now we will create figure and axis object
# Axes is technically a numpy array

figure,axes=plt.subplots(2,2)
```

```python
#now we have 2 rows and 2 columns .we will fill each of these individual subplots.
#we have selected the subplot at row 0, column 0. so only this one is filled in
x= np.array([1,2,3,4,5])
figure,axes=plt.subplots(2,2)
axes[0,0].plot(x,x**2,color="red")
axes[0,0].set_title("x**2")        # setting the title for the subplot
axes[0,0].set_xlabel("x-axis 1")   # setting the xlabel for the subplot
axes[0,0].set_ylabel("y-axis 1")   # setting the ylabel for the subplot
plt.tight_layout()                 #to prevent overlapping
plt.show()
              
```

```python
x= np.array([1,2,3,4,5])
figure,axes=plt.subplots(2,2)
axes[0,0].plot(x,x**2,color="red")
axes[0,0].set_title("x**2")         
axes[0,0].set_xlabel("x-axis 1")
axes[0,0].set_ylabel("y-axis 1")

axes[0,1].plot(x,x**3,color="blue")
axes[0,1].set_title("x**3")
axes[0,1].set_xlabel("x-axis 2")
axes[0,1].set_ylabel("y-axis 2")
plt.tight_layout()
plt.show()
```
